```python
level = int(input("경보 레벨? "))
if level >= 10:
    print("심각")
elif level >= 7:
    print("주의")
else:
    print("정상")
```

```python
ip = "1.1.1.1"
level = 10
if level >= 10 and ip != "127.0.0.1":
    print("외부 심각 경보:", ip)
else:
    print("내부 또는 경미")
```

```python
level = 10
ip = "1.1.1.1"
if level >= 7:
    if ip != "127.0.0.1":
        print("외부 경보 — 즉시 대응")
    else:
        print("내부 경보")
else:
    print("정상")
```
